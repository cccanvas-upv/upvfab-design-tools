## 2. Interference mechanism and initial geometry

**SYMMETRIC INTERFERENCE**
- Input position: y = 0
- Output positions: y = ?1.20 ?m
- Edge gap between output tapers at the MMI face: 1.10 ?m

The initial geometry comprises the MMI region, one centered input taper, two output tapers, and straight access waveguides. The polygons are generated in the [reproduction notebook](../../work-in-progress/cccanvas-upv/mmi-1x2-soi220-5050/mmi1x2_soi220_5050_reproduction.ipynb), using the FDTD geometry plotting and conversion utilities of `upvfab_design_tools`.

## 3. Design Flow

1. **Geometry selection:** the source notebook specifies the MMI dimensions and symmetric access positions.
2. **FDTD setup:** the geometry, source and monitors are built in Tidy3D.
3. **Simulation and analysis:** the output modal amplitudes and field are used to calculate the splitting ratio, excess loss and phase over the wavelength sweep.

An analytical beat-length calculation or optimization sweep is not documented in the source notebook.
