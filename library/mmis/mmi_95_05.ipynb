# MMI 2x2 (95:05)

## 1. Device Overview

This notebook presents the design and simulation results of a 2x2 multimode interference (MMI) coupler, designed to achieve a power splitting ratio of 95:05 and a  90º  relative phase difference between output ports. 

Note that 95:05 does not correspond to a canonical ratio. 95% of the input optical power is directed to the cross port and 5% to the bar port.


| Parameter | Specification |
|---|---|
| Device | 2x2  MMI coupler |
| Target splitting ratio | 95:05|
| Target output phase difference | 90º |
| Operating wavelength | $\lambda_0 = 1.55\ \mu m$ |
| Operating band | C-band |
| Polarization | TE |
| Waveguide core | SiN (300 nm thickness) |
| Cladding | SiO₂ |

## 2. Interference mechanism and initial geometry

To achieve this ratio, a Symmetric Type B Butterfly geometry must be implemented. Its equations are shown below: 

**SYMMETRIC BUTTERFLY GEOMETRY EQUATIONS**
$$L_{0} = \delta_W \frac{1}{3}(3L_{\pi})$$

$$\delta_W = 1 - {2\Delta W}/{W_{\mathrm{MMI}}}$$

$$\Delta W ={d_{\Omega}W_{\mathrm{MMI}}}/{2}$$

$$d_{\Omega} = {2}\arccos\left(\sqrt{0.95}\right)/{\pi}$$


- Input positions: +-$W_{mmi}/6$
- Output positions: +-$W_{mmi}/6$ 

## 3. Design Flow

The device was designed following a **conventional** MMI design flow:

1. **Width selection:** $W_{\mathrm{MMI}} = 10\ \mu m$.

2. **Analytical design:** the beat length ($L_\pi$) was calculated using the propagation constants obtained from the mode solvers.

3. **Parametric FDTD optimization:** varying the MMI length, $\Delta W$,  the position of the access waveguides, and the taper widths and lengths.
