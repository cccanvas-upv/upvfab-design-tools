# MMI 2x2 (W=10)

## 1. Device Overview

This notebook presents the design and simulation results of a 2x2 multimode interference (MMI) coupler, designed to achieve a power splitting ratio of (50:50) and a  90º relative phase difference between output ports.

| Parameter | Specification |
|---|---|
| Device | 2x2 MMI coupler |
| Target splitting ratio | 50:50|
| Target output phase difference | 90º|
| Operating wavelength | $\lambda_0 = 1.55\ \mu m$ |
| Operating band | C-band |
| Polarization | TE |
| Waveguide core | SiN (300 nm thickness) |
| Cladding | SiO₂ |

## 2. Interference mechanism and initial geometry

**PAIRED INTERFERENCE**
- $L_{mmi}$ = ${L_{\pi}}/2$
- Input positionS: +-$W_{mmi}/6$ 
- Output positions: +-$W_{mmi}/6$ 

The initial device geometry is generated using the `mmi_2X2_vertices()` function, which constructs the polygons defining the MMI region, access waveguides, and input/output tapers.
The function is imported from `upvfab_design_tools.tidy3d_plugin.fdtd`. 


## 3. Design Flow

The device was designed following a **conventional** MMI design flow:

1. **Width selection:** $W_{\mathrm{MMI}} = 10\ \mu m$.

2. **Analytical design:** the beat length ($L_\pi$) was calculated using the propagation constants obtained from the mode solvers.

3. **Initial EME optimization:** the initial geometry was evaluated and optimized using the Eigenmode Expansion (EME) method.

4. **Parametric FDTD optimization:** varying the MMI length, the position of the access waveguides, and the taper widths and lengths (if necessary).
