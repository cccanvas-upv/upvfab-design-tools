# MMI 2x2 SOI 220 nm (50:50)

## 1. Device Overview

| Parameter | Specification |
|---|---|
| Device | 2x2 MMI coupler |
| Target splitting ratio | 50:50 |
| Target output phase difference | 90 degrees |
| Operating wavelength | 1.55 um |
| Band | 1.50-1.60 um |
| Core | Silicon, 220 nm |
| Cladding | Homogeneous SiO2 |
| Excited input | Upper input |

## 2. Interference mechanism and initial geometry

**Paired interference**

The upper input excites the multimode section; the two outputs are symmetrically positioned at y = +/-0.8 um.

The MMI and access waveguide polygons are built using `mmi_2x2_vertices()` from `upvfab_design_tools.tidy3d_plugin`.

## 3. Design Flow

1. Select MMI and access geometry from the source design.
2. Build the 2x2 FDTD simulation with the package.
3. Excite the upper input and evaluate output modes and the wavelength sweep.

The original repository includes a separate length sweep; this record uses its specified 26.8 um geometry.

### Sources and Monitors

The upper input is excited. Mode and flux monitors are placed at each output, and a field monitor records the core mid-plane. The package source and monitor settings differ from the original notebook. The solver reached the field-decay shutoff threshold.

The simulation is built using `build_mmi_2x2_fdtd_simulation()`; see the linked WIP notebook for the parameters and saved-data loading.

## 7. Wavelength sweep results



The 50:50 tolerance is +/-0.01 in coupling ratio. The nominal point does not satisfy that criterion, so the reported continuous bandwidth around 1.55 um is 0 nm.
