# MMI 2x4 (90º hybrid)

## 1. Device Overview

This notebook presents the design and simulation results of a 2x4 multimode interference (MMI) coupler, designed to achieve a power splitting ratio of (25:25:25:25) and to work as a 90º hybrid. Ideally, the four outputs exhibit relative phase offsets equivalent to 0º, 90º, 180º, and 270º, forming two complementary output pairs in quadrature.

| Parameter | Specification |
|---|---|
| Device | 2x4 MMI coupler |
| Target splitting ratio |25:25:25:25|
| Target output phase difference (HYBRID) | 90º, 180º, 0º, 270º (-90º) |
| Operating wavelength | $\lambda_0 = 1.55\ \mu m$ |
| Operating band | C-band |
| Polarization | TE |
| Waveguide core | SiN (300 nm thickness) |
| Cladding | SiO₂ |

## 2. Interference mechanism and initial geometry

**"GENERAL INTERFERENCE"**
The equations used in this design are the ones corresponing to a 4x4 geometry: 
- $L_{mmi}$ = $3{L_{\pi}}/4$
- Input position: +$W_{mmi}/8$ , -$W_{mmi}/8$ 
- Output positions: +-$W_{mmi}/8$ , +-$W_{mmi}/8$ 

The initial device geometry is generated using the `mmi_2x4_vertices()` function, which constructs the polygons defining the MMI region, access waveguides, and input/output tapers.
The function is imported from `upvfab_design_tools.tidy3d_plugin.fdtd`. 


## 3. Design Flow

The device was designed following a **conventional** MMI design flow:

1. **Width selection:** $W_{\mathrm{MMI}} = 16\ \mu m$.

2. **Analytical design:** the beat length ($L_\pi$) was calculated using the propagation constants obtained from the mode solvers.

3. **Initial EME optimization:** the initial geometry was evaluated and optimized using the Eigenmode Expansion (EME) method.

4. **Parametric FDTD optimization (SINGLE INPUT):** varying the MMI length, the position of the access waveguides, and the taper widths and lengths (if necessary).

5. **HYBRID FUNCTIONALITY CHECK:** after the single input optimization, simulations with hybrid excitation are needed to verify its performance.
