# MMI 1x4

## 1. Device Overview

This notebook presents the design and simulation results of a 1×4 multimode interference (MMI) coupler, designed to achieve a power splitting ratio of (25:25:25:25) and a (0°, 90º, 90º, 0º) relative phase difference between output ports.

| Parameter | Specification |
|---|---|
| Device | 1×4 MMI coupler |
| Target splitting ratio | 25:25:25:25|
| Target output phase difference | 0°, 90º, 90º, 0º |
| Operating wavelength | $\lambda_0 = 1.55\ \mu m$ |
| Operating band | C-band |
| Polarization | TE |
| Waveguide core | SiN (300 nm thickness) |
| Cladding | SiO₂ |

## 2. Interference mechanism and initial geometry

**SYMMETRIC INTERFERENCE**
- $L_{mmi}$ = $3{L_{\pi}}/16$
- Input position: y = 0 
- Output positions: +-3 $W_{mmi}/8$ , +-$W_{mmi}/8$ 

The initial device geometry is generated using the `mmi_1x4_vertices()` function, which constructs the polygons defining the MMI region, access waveguides, and input/output tapers.
The function is imported from `upvfab_design_tools.tidy3d_plugin.fdtd`. 


## 3. Design Flow

The device was designed following a **conventional** MMI design flow:

1. **Width selection:** $W_{\mathrm{MMI}} = 12\ \mu m$.

2. **Analytical design:** he beat length ($L_\pi$) was calculated using the propagation constants obtained from the mode solvers.

3. **Initial EME optimization:** the initial geometry was evaluated and optimized using the Eigenmode Expansion (EME) method.

4. **Parametric FDTD optimization:** varying the MMI length, the position of the access waveguides, and the taper widths and lengths (if necessary).
